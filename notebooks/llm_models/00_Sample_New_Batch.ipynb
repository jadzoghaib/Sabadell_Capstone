# 00 — Sample New Batch

Lightweight notebook that samples a **new, non-overlapping 100-loan batch** from the test set and saves it as a CSV.  
Run this notebook **once**, then close it to free RAM before running `04c_Robustness.ipynb`.

```python
from llm_utils import sample_new_batch, RESULTS_DIR
import os
```

```python
# Sample a new non-overlapping batch of 100 loans (with descriptions)
new_sample = sample_new_batch(n=100, random_state=99)

print(f"New sample: {new_sample.shape[0]} loans, {new_sample.shape[1]} columns")
print(f"\nClass distribution:")
print(new_sample['loan_status'].value_counts())
print(f"\nHas description: {new_sample['desc'].notna().all()}")
```

```python
# Save to CSV
os.makedirs(RESULTS_DIR, exist_ok=True)
out_path = f"{RESULTS_DIR}/04c_new_batch_sample.csv"
new_sample.to_csv(out_path, index=False)
print(f"Saved to {out_path}")
print("\nYou can now close this notebook to free RAM, then run 04c_Robustness.ipynb.")
```
